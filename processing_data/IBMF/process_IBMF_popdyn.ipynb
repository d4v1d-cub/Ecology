```python
import numpy as np
import os
import fnmatch
```

```python
def filter_files(path, pattern):
    files_mu_sigma = []
    # Iterate through the files in the specified directory
    for filename in os.listdir(path):
        if fnmatch.fnmatch(filename, pattern):
            parts = filename.split('_')
            files_mu_sigma.append((filename, float(parts[22]), float(parts[24]), int(parts[26])))
    sorted_pairs = sorted(files_mu_sigma, key=lambda x: (x[1], x[2], x[3]))  # Sort by mu value, sigma value, and S value
    return sorted_pairs
```

```python
def average_last_data(path, filename, start_from=1000, pos_m=1, pos_delta=4):
    fin = open(f'{path}/{filename}', 'r')
    j = fin.readline()
    while j[0] == '#':
        j = fin.readline()
        if not j:
            break
    av_m = 0.0
    av_m_sqr = 0.0
    av_delta = 0.0
    av_delta_sqr = 0.0
    counter = 0
    line = j.split()
    iteration = int(line[0])
    if iteration > start_from:
        av_m += float(line[pos_m])
        av_m_sqr += float(line[pos_m])**2
        av_delta += float(line[pos_delta])
        av_delta_sqr += float(line[pos_delta])**2
        counter += 1
    while True:
        j = fin.readline()
        if not j:
            break
        line = j.split()
        iteration = int(line[0])
        if iteration > start_from:
            av_m += float(line[pos_m])
            av_m_sqr += float(line[pos_m])**2
            av_delta += float(line[pos_delta])
            av_delta_sqr += float(line[pos_delta])**2
            counter += 1
    if counter > 0:
        av_m /= counter
        av_m_sqr /= counter
        av_delta /= counter
        av_delta_sqr /= counter
        error_m = np.sqrt((av_m_sqr - av_m**2) / counter)
        error_delta = np.sqrt((av_delta_sqr - av_delta**2) / counter)
    else:
        error_m = 0.0
        error_delta = 0.0
    
    fin.close()
    return av_m, error_m, av_delta, error_delta
```

```python
def all_averages(pathout, fileout, path, pattern):
    files = filter_files(path, pattern)
    fout = open(f'{pathout}/{fileout}', 'w')
    fout.write('#mu(X10^3) sigma(X10^3) S av_m error_m av_delta error_delta\n')
    for filename, mu, sigma, S in files:
        av_m, error_m, av_delta, error_delta = average_last_data(path, filename)
        fout.write(f'{int(mu*1000)} {int(sigma*1000)} {S} {av_m} {error_m} {av_delta} {error_delta}\n')
        print(f'Processed file: mu={mu}, sigma={sigma}, S={S}')
    fout.close()
```

```python
def paste_all_files(pathout, fileout, path, pattern, ndigits=3):
    files = filter_files(path, pattern)
    fout = open(f'{pathout}/{fileout}', 'w')
    fout.write('#mu sigma S niter convergence av(m) error(m) av(dead) seed_initcond runtime\n')
    for filename, mu, sigma, S in files:
        fin = open(f'{path}/{filename}', 'r')
        line = fin.readline()
        fin.close()
        if not line:
            continue
        fout.write(f'{mu:.{ndigits}f}\t{sigma:.{ndigits}f}\t{S}\t{line}')
        print(f'Processed file: mu={mu}, sigma={sigma}, S={S}')
    fout.close()
```
